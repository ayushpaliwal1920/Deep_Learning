```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# generate data :

from sklearn.datasets import make_regression

x, y = make_regression(
    n_samples=100,
    n_features=2,
    noise=10,
    random_state=42
)

print(x.shape, y.shape)
```

```python
# convert to DataFrame
x = pd.DataFrame(x, columns=['x1', 'x2'])
y = pd.Series(y)   #
```

```python
# plot
plt.scatter(x.iloc[:, 0], x.iloc[:, 1], c=y)
plt.show()
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()

model.add(Dense(10,activation='sigmoid',input_dim =2))
model.add(Dense(1,activation="sigmoid"))

model.summary()
```

```python
#  default values :

model.get_weights()
```

```python
inital_weights = model.get_weights()
```

```python
# set weights to 0

inital_weights[0] = np.ones(model.get_weights()[0].shape)*0.5
inital_weights[1] = np.ones(model.get_weights()[1].shape)*0.5
inital_weights[2] = np.ones(model.get_weights()[2].shape)*0.5
inital_weights[3] = np.ones(model.get_weights()[3].shape)*0.5
```

```python
model.set_weights(inital_weights)
```

```python
model.get_weights()
```

```python
model.compile(loss="binary_crossentropy",optimizer = "adam",metrics = ['accuracy'])
```

```python
history = model.fit(x,y,epochs=100,validation_split=0.2)
```

```python
model.get_weights()
```

```python
from mlxtend.plotting import plot_decision_regions

plot_decision_regions(X=x.values,
                      y=y.astype(int).values,
                      clf=model,
                      legend=2)
```
