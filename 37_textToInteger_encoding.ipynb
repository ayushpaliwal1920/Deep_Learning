```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense , SimpleRNN
import numpy as np
```

# Input data prepration :

```python
docs = ['go india',
        'india india',
        'hip hip hurray',
        'jeetega bhai jeetaga india jeetaga',
        'bharat mata ki jai',
        'kohli kohli',
        'sachin sachin',
        'dhoni dhoni',
        'modi ji ki jai',
        'inquilab zindabad']
```

```python
# Tokenization :

from tensorflow.keras.preprocessing.text import Tokenizer

tokenizer = Tokenizer(oov_token='<nothing>') # oov_token : placeholder for when data do not consist the word
```

```python
tokenizer.fit_on_texts(docs) # tokenization on docs
```

```python
tokenizer.word_index # print uniue words and their indexes
```

```python
tokenizer.word_counts # frequency count of words
```

```python
tokenizer.document_count # total sentences
```

```python
sequences = tokenizer.texts_to_sequences(docs) # replaces sentences with their token numbers

sequences
```

```python
## padding :

from tensorflow.keras.utils import pad_sequences

sequences = pad_sequences(sequences , padding = 'post') # post : zeros will be added at the last , pre : zeroes will be added at the starting
```
