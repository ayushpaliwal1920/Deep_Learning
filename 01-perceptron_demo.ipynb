```python
import pandas as pd 
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Placement.csv")
df.head()
```

```python
print(df.shape)
```

```python
sns.scatterplot(x="CGPA",y="IQ",data=df,hue="Placement")
plt.show()
```

Scalling : 

```python
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()

ss.fit(df[["IQ"]])
df["IQ_scaled"] = pd.DataFrame(ss.transform(df[["IQ"]]))
```

```python
df
```

```python
x = df[["CGPA" , "IQ_scaled"]]
y = df["Placement"]
```

```python
from sklearn.linear_model import Perceptron
```

```python
p = Perceptron()
p.fit(x,y)
```

```python
p.coef_  # value of w1,w1 means slopes of  a linear equation
```

```python
p.intercept_
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(x.values,y.values,clf=p)
plt.show()
```
