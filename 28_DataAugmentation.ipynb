```python
# importing libraries

import tensorflow as tf
from tensorflow import keras
from keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
img = image.load_img('/content/cat.jpeg', target_size=(200,200))
```

```python
import matplotlib.pyplot as plt
```

```python
img
```

```python
# data Augmentation :

datagen = ImageDataGenerator(
    rotation_range = 30, # angle
    shear_range = 0.2, # distortion
    zoom_range = 0.2, # zoom
    horizontal_flip = True,
    vertical_flip = False,
    width_shift_range = 0.2,
    height_shift_range = 0.2,
    fill_mode = 'nearest'
)
```

```python
# converting img in numpy array

img = image.img_to_array(img)
```

```python
img.shape
```

```python
input_batch = img.reshape(1,200,200,3)
```

```python
i = 0

# 'flow' is used for one image and 'flow_from_dir' is used when we are using a image folder or directory

for output in datagen.flow(input_batch,batch_size = 1,save_to_dir = 'aug'):
    i = i+1

    if i == 10:
      break
```
